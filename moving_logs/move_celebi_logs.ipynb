```python
import os
from pathlib import Path
```

```python
ROOT_0 = Path('./logs/use_tuned_vanilla_hp_run_ea/')
ROOT_1 = Path('./logs_celebi/use_tuned_vanilla_hp_run_ea/')
exps = [dir for dir in ROOT_1.iterdir() if dir.is_dir()]
DATASETS = ['german', 'bail', 'pokec_z', 'pokec_n', 'nba']
TARGET_BACKBONE = 'gat'
for exp in exps:
    for DATASET in DATASETS:
        path_0 = ROOT_0 / exp.name / DATASET / TARGET_BACKBONE
        path_1 = ROOT_1 / exp.name / DATASET / TARGET_BACKBONE
        if not path_0.exists():
            os.rename(path_1, path_0)
            print(f'Moved {path_1} to {path_0}')
```
